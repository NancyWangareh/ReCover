```python
import json
import time

output_dir = data_dir / "parsed_output"
output_dir.mkdir(parents=True, exist_ok=True)

parsing_results = {}
failed_files = []

for pdf_path in pdf_files:

    print(f"\n{'=' * 60}")
    print(f"Processing: {pdf_path.name}")
    print(f"{'=' * 60}")

    try:
        # Upload the PDF
        uploaded_file = client.files.create(
            file=str(pdf_path),
            purpose="parse"
        )

        print("Upload successful. Starting parsing...")

        # Parse using the v2 API
        result = client.parsing.parse(
            file_id=uploaded_file.id,
            tier="agentic",
            version="latest",
            expand=["markdown"]
        )

        # Extract page content
        extracted_pages = []

        for page in result.markdown.pages:
            if page.markdown and page.markdown.strip():
                extracted_pages.append({
                    "page_number": page.page_number,
                    "text": page.markdown
                })

        # Combine pages into one Markdown document
        combined_text = "\n\n".join(
            f"<!-- Page {page['page_number']} -->\n\n{page['text']}"
            for page in extracted_pages
        )

        # Save individual document as Markdown
        output_path = output_dir / f"{pdf_path.stem}.md"

        output_path.write_text(
            combined_text,
            encoding="utf-8"
        )

        parsing_results[pdf_path.name] = {
            "status": "success",
            "file_id": uploaded_file.id,
            "pages_extracted": len(extracted_pages),
            "characters_extracted": len(combined_text),
            "output_file": str(output_path)
        }

        print(f"Success: {len(extracted_pages)} pages extracted.")
        print(f"Characters extracted: {len(combined_text):,}")
        print(f"Saved to: {output_path}")

    except Exception as e:
        failed_files.append(pdf_path.name)

        parsing_results[pdf_path.name] = {
            "status": "failed",
            "error_type": type(e).__name__,
            "error": str(e)
        }

        print(f"FAILED: {type(e).__name__}: {e}")

    # Brief pause between documents
    time.sleep(1)

# Save processing report
report_path = output_dir / "parsing_report.json"

report_path.write_text(
    json.dumps(parsing_results, indent=2),
    encoding="utf-8"
)

print("\n" + "=" * 60)
print("PARSING SUMMARY")
print("=" * 60)
print("Total PDFs:", len(pdf_files))
print("Successful:", sum(
    1 for item in parsing_results.values()
    if item["status"] == "success"
))
print("Failed:", len(failed_files))

if failed_files:
    print("\nFiles requiring attention:")
    for filename in failed_files:
        print("-", filename)

print("\nOutput directory:", output_dir)
```

## Create chunks with source metadata

```python
from pathlib import Path
from langchain_text_splitters import RecursiveCharacterTextSplitter

data_dir = Path("/content/drive/MyDrive/ReCover/data")
parsed_dir = data_dir / "parsed_output"

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = []

for file in sorted(parsed_dir.glob("*.md")):
    text = file.read_text(encoding="utf-8")

    if not text.strip():
        print(f"Skipping empty document: {file.name}")
        continue

    document_chunks = splitter.split_text(text)

    for index, chunk_text in enumerate(document_chunks):
        chunks.append({
            "chunk_id": f"{file.stem}_{index:04d}",
            "source": file.name,
            "chunk_index": index,
            "text": chunk_text
        })

    print(f"{file.name}: {len(document_chunks)} chunks")

print(f"\nTotal chunks created: {len(chunks)}")
```

### Load Chunks

```python
from pathlib import Path
from langchain_text_splitters import RecursiveCharacterTextSplitter

data_dir = Path("/content/drive/MyDrive/ReCover/data")
parsed_dir = data_dir / "parsed_output"

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = []

for file in sorted(parsed_dir.glob("*.md")):
    text = file.read_text(encoding="utf-8")

    if not text.strip():
        continue

    for index, chunk_text in enumerate(splitter.split_text(text)):
        chunks.append({
            "chunk_id": f"{file.stem}_{index:04d}",
            "source": file.name,
            "chunk_index": index,
            "text": chunk_text
        })

print("Total chunks:", len(chunks))
```

### Initialize the Embedding Model

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")
```

### Creating a Chroma DB Collection

```python
import chromadb

chroma_dir = data_dir / "chroma_db"
chroma_dir.mkdir(parents=True, exist_ok=True)

chroma_client = chromadb.PersistentClient(
    path=str(chroma_dir)
)

collection = chroma_client.get_or_create_collection(
    name="ReCover",
    metadata={"hnsw:space": "cosine"}
)

print("ChromaDB collection initialized.")
print("Collection:", collection.name)
print("Existing records:", collection.count())
```

### Generating embeddings and Inserting chunks

```python
import hashlib

batch_size = 32

for start in range(0, len(chunks), batch_size):
    batch = chunks[start:start + batch_size]

    ids = [item["chunk_id"] for item in batch]
    texts = [item["text"] for item in batch]

    embeddings = embedding_model.encode(
        texts,
        normalize_embeddings=True,
        show_progress_bar=False
    ).tolist()

    metadatas = [
        {
            "source": item["source"],
            "chunk_index": item["chunk_index"],
            "text_hash": hashlib.sha256(
                item["text"].encode("utf-8")
            ).hexdigest()
        }
        for item in batch
    ]

    collection.upsert(
        ids=ids,
        documents=texts,
        embeddings=embeddings,
        metadatas=metadatas
    )

    print(
        f"Indexed {min(start + batch_size, len(chunks))}"
        f"/{len(chunks)} chunks"
    )

print("\nIndexing complete!")
print("Total indexed records:", collection.count())
```

## Test Semantic Retrieval

### Creating a search function

```python
def retrieve_documents(query, top_k=5):
    query_embedding = embedding_model.encode(
        query,
        normalize_embeddings=True
    ).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=min(top_k, collection.count()),
        include=["documents", "metadatas", "distances"]
    )

    retrieved = []

    for i in range(len(results["ids"][0])):
        retrieved.append({
            "chunk_id": results["ids"][0][i],
            "text": results["documents"][0][i],
            "metadata": results["metadatas"][0][i],
            "distance": results["distances"][0][i]
        })

    return retrieved
```

```python
%%writefile /content/drive/MyDrive/ReCover/src/llm_client.py

import os
from google import genai
from google.genai import types

client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)


def generate_answer(question, context):
    """
    Generate a grounded answer using Gemini and retrieved evidence.
    """

    system_prompt = """
    You are ReCover, a microinsurance information assistant
    for users in Kenya.

    Answer using only the supplied document excerpts.

    For product questions:
    - Prioritize product names and providers.
    - Explain coverage where the evidence supports it.
    - Include exclusions, eligibility and premiums when available.
    - Do not invent missing information.
    - Cite the source document names.
    - If the evidence is insufficient, say so.
    - Prefer concise lists or tables over long background summaries.
    """

    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=f"""
        Question:
        {question}

        Retrieved document excerpts:
        {context}
        """,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt,
            temperature=0.2,
            max_output_tokens=1200
        )
    )

    if not response.text:
        raise RuntimeError("Gemini returned an empty response.")

    return response.text

```

```python
%%writefile /content/drive/MyDrive/ReCover/src/rag_pipeline.py

from vector_store import retrieve_documents
from llm_client import generate_answer


def ask_recover(question, n_results=5):
    """
    Retrieve relevant documents and generate an answer.
    """

    results = retrieve_documents(
        question,
        n_results=n_results
    )

    documents = results["documents"][0]
    metadatas = results["metadatas"][0]

    context_parts = []

    for i, document in enumerate(documents):
        metadata = metadatas[i] or {}
        source = metadata.get("source", "Unknown source")

        context_parts.append(
            f"Source: {source}\n"
            f"Excerpt:\n{document}"
        )

    context = "\n\n".join(context_parts)

    if not context.strip():
        return "I could not find relevant information in the knowledge base."

    return generate_answer(
        question=question,
        context=context
    )
```
