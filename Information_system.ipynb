```python
import json
import time

output_dir = data_dir / "parsed_output"
output_dir.mkdir(parents=True, exist_ok=True)

parsing_results = {}
failed_files = []

for pdf_path in pdf_files:

    print(f"\n{'=' * 60}")
    print(f"Processing: {pdf_path.name}")
    print(f"{'=' * 60}")

    try:
        # Upload the PDF
        uploaded_file = client.files.create(
            file=str(pdf_path),
            purpose="parse"
        )

        print("Upload successful. Starting parsing...")

        # Parse using the v2 API
        result = client.parsing.parse(
            file_id=uploaded_file.id,
            tier="agentic",
            version="latest",
            expand=["markdown"]
        )

        # Extract page content
        extracted_pages = []

        for page in result.markdown.pages:
            if page.markdown and page.markdown.strip():
                extracted_pages.append({
                    "page_number": page.page_number,
                    "text": page.markdown
                })

        # Combine pages into one Markdown document
        combined_text = "\n\n".join(
            f"<!-- Page {page['page_number']} -->\n\n{page['text']}"
            for page in extracted_pages
        )

        # Save individual document as Markdown
        output_path = output_dir / f"{pdf_path.stem}.md"

        output_path.write_text(
            combined_text,
            encoding="utf-8"
        )

        parsing_results[pdf_path.name] = {
            "status": "success",
            "file_id": uploaded_file.id,
            "pages_extracted": len(extracted_pages),
            "characters_extracted": len(combined_text),
            "output_file": str(output_path)
        }

        print(f"Success: {len(extracted_pages)} pages extracted.")
        print(f"Characters extracted: {len(combined_text):,}")
        print(f"Saved to: {output_path}")

    except Exception as e:
        failed_files.append(pdf_path.name)

        parsing_results[pdf_path.name] = {
            "status": "failed",
            "error_type": type(e).__name__,
            "error": str(e)
        }

        print(f"FAILED: {type(e).__name__}: {e}")

    # Brief pause between documents
    time.sleep(1)

# Save processing report
report_path = output_dir / "parsing_report.json"

report_path.write_text(
    json.dumps(parsing_results, indent=2),
    encoding="utf-8"
)

print("\n" + "=" * 60)
print("PARSING SUMMARY")
print("=" * 60)
print("Total PDFs:", len(pdf_files))
print("Successful:", sum(
    1 for item in parsing_results.values()
    if item["status"] == "success"
))
print("Failed:", len(failed_files))

if failed_files:
    print("\nFiles requiring attention:")
    for filename in failed_files:
        print("-", filename)

print("\nOutput directory:", output_dir)
```

## Create chunks with source metadata

```python


from pathlib import Path
from langchain_text_splitters import RecursiveCharacterTextSplitter

data_dir = Path("/content/drive/MyDrive/ReCover/data")
parsed_dir = data_dir / "parsed_output"

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = []

for file in sorted(parsed_dir.glob("*.md")):
    text = file.read_text(encoding="utf-8")

    if not text.strip():
        print(f"Skipping empty document: {file.name}")
        continue

    document_chunks = splitter.split_text(text)

    for index, chunk_text in enumerate(document_chunks):
        chunks.append({
            "chunk_id": f"{file.stem}_{index:04d}",
            "source": file.name,
            "chunk_index": index,
            "text": chunk_text
        })

    print(f"{file.name}: {len(document_chunks)} chunks")

print(f"\nTotal chunks created: {len(chunks)}")
```

### Load Chunks

```python
from pathlib import Path
from langchain_text_splitters import RecursiveCharacterTextSplitter

data_dir = Path("/content/drive/MyDrive/ReCover/data")
parsed_dir = data_dir / "parsed_output"

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = []

for file in sorted(parsed_dir.glob("*.md")):
    text = file.read_text(encoding="utf-8")

    if not text.strip():
        continue

    for index, chunk_text in enumerate(splitter.split_text(text)):
        chunks.append({
            "chunk_id": f"{file.stem}_{index:04d}",
            "source": file.name,
            "chunk_index": index,
            "text": chunk_text
        })

print("Total chunks:", len(chunks))
```

### Initialize the Embedding Model

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")
```

### Creating a Chroma DB Collection

```python
import chromadb

chroma_dir = data_dir / "chroma_db"
chroma_dir.mkdir(parents=True, exist_ok=True)

chroma_client = chromadb.PersistentClient(
    path=str(chroma_dir)
)

collection = chroma_client.get_or_create_collection(
    name="ReCover",
    metadata={"hnsw:space": "cosine"}
)

print("ChromaDB collection initialized.")
print("Collection:", collection.name)
print("Existing records:", collection.count())
```

### Generating embeddings and Inserting chunks

```python
import hashlib

batch_size = 32

for start in range(0, len(chunks), batch_size):
    batch = chunks[start:start + batch_size]

    ids = [item["chunk_id"] for item in batch]
    texts = [item["text"] for item in batch]

    embeddings = embedding_model.encode(
        texts,
        normalize_embeddings=True,
        show_progress_bar=False
    ).tolist()

    metadatas = [
        {
            "source": item["source"],
            "chunk_index": item["chunk_index"],
            "text_hash": hashlib.sha256(
                item["text"].encode("utf-8")
            ).hexdigest()
        }
        for item in batch
    ]

    collection.upsert(
        ids=ids,
        documents=texts,
        embeddings=embeddings,
        metadatas=metadatas
    )

    print(
        f"Indexed {min(start + batch_size, len(chunks))}"
        f"/{len(chunks)} chunks"
    )

print("\nIndexing complete!")
print("Total indexed records:", collection.count())
```

## Test Semantic Retrieval

### Creating a search function

```python
def retrieve_documents(query, top_k=5):
    query_embedding = embedding_model.encode(
        query,
        normalize_embeddings=True
    ).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=min(top_k, collection.count()),
        include=["documents", "metadatas", "distances"]
    )

    retrieved = []

    for i in range(len(results["ids"][0])):
        retrieved.append({
            "chunk_id": results["ids"][0][i],
            "text": results["documents"][0][i],
            "metadata": results["metadatas"][0][i],
            "distance": results["distances"][0][i]
        })

    return retrieved
```

```python
%%writefile /content/drive/MyDrive/ReCover/src/llm_client.py

import os
from google import genai
from google.genai import types

client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)


def generate_answer(question, context):
    """
    Generate a grounded answer using Gemini and retrieved evidence.
    """

    system_prompt = """
    You are ReCover, a microinsurance information assistant
    for users in Kenya.

    Answer using only the supplied document excerpts.

    For product questions:
    - Prioritize product names and providers.
    - Explain coverage where the evidence supports it.
    - Include exclusions, eligibility and premiums when available.
    - Do not invent missing information.
    - Cite the source document names.
    - If the evidence is insufficient, say so.
    - Prefer concise lists or tables over long background summaries.
    """

    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=f"""
        Question:
        {question}

        Retrieved document excerpts:
        {context}
        """,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt,
            temperature=0.2,
            max_output_tokens=1200
        )
    )

    if not response.text:
        raise RuntimeError("Gemini returned an empty response.")

    return response.text

```

#### rag_pipeline.py file

```python
%%writefile /content/drive/MyDrive/ReCover/src/rag_pipeline.py

from vector_store import retrieve_documents
from llm_client import generate_answer


def ask_recover(question, n_results=5):
    """
    Retrieve relevant documents and generate an answer.
    """

    results = retrieve_documents(
        question,
        n_results=n_results
    )

    documents = results["documents"][0]
    metadatas = results["metadatas"][0]

    context_parts = []

    for i, document in enumerate(documents):
        metadata = metadatas[i] or {}
        source = metadata.get("source", "Unknown source")

        context_parts.append(
            f"Source: {source}\n"
            f"Excerpt:\n{document}"
        )

    context = "\n\n".join(context_parts)

    if not context.strip():
        return "I could not find relevant information in the knowledge base."

    return generate_answer(
        question=question,
        context=context
    )
```

```python
%%writefile /content/drive/MyDrive/ReCover/src/document_parser.py

"""
Parse PDF documents using LlamaCloud and save the output as Markdown.
"""

import os
from pathlib import Path

from llama_cloud_services import LlamaParse


PROJECT_DIR = Path("/content/drive/MyDrive/ReCover")
INPUT_DIR = PROJECT_DIR / "data" / "original"
OUTPUT_DIR = PROJECT_DIR / "data" / "parsed_output"


def parse_documents(input_dir=None, output_dir=None):
    """
    Parse PDFs in the input directory and save Markdown output.

    The LLAMA_CLOUD_API_KEY environment variable must be configured.

    Returns:
        list[Path]: Paths to the saved Markdown files.
    """
    input_dir = Path(input_dir or INPUT_DIR)
    output_dir = Path(output_dir or OUTPUT_DIR)

    if not os.getenv("LLAMA_CLOUD_API_KEY"):
        raise EnvironmentError(
            "LLAMA_CLOUD_API_KEY is not configured."
        )

    if not input_dir.exists():
        raise FileNotFoundError(
            f"Input directory does not exist: {input_dir}"
        )

    output_dir.mkdir(parents=True, exist_ok=True)

    parser = LlamaParse(
        api_key=os.environ["LLAMA_CLOUD_API_KEY"],
        result_type="markdown",
    )

    saved_files = []

    for pdf_path in sorted(input_dir.glob("*.pdf")):
        output_path = output_dir / f"{pdf_path.stem}.md"

        if output_path.exists():
            print(f"Skipping existing file: {output_path.name}")
            saved_files.append(output_path)
            continue

        print(f"Parsing: {pdf_path.name}")

        documents = parser.parse(str(pdf_path))

        markdown_parts = []

        for document in documents:
            text = getattr(document, "text", "")
            if text:
                markdown_parts.append(text)

        if not markdown_parts:
            print(f"Warning: No text extracted from {pdf_path.name}")
            continue

        output_path.write_text(
            "\n\n".join(markdown_parts),
            encoding="utf-8",
        )

        saved_files.append(output_path)
        print(f"Saved: {output_path}")

    print(f"\nParsing complete. Markdown files available: {len(saved_files)}")

    return saved_files
```

chunking.py file

```python
%%writefile /content/drive/MyDrive/ReCover/src/chunking.py

"""
Split parsed Markdown documents into overlapping text chunks.
"""

from pathlib import Path
from langchain_text_splitters import RecursiveCharacterTextSplitter


PROJECT_DIR = Path("/content/drive/MyDrive/ReCover")
PARSED_DIR = PROJECT_DIR / "data" / "parsed_output"


def chunk_text(
    text,
    chunk_size=1000,
    chunk_overlap=150,
):
    """
    Split a text string into overlapping chunks.

    Args:
        text: Text to split.
        chunk_size: Maximum chunk size in characters.
        chunk_overlap: Number of overlapping characters.

    Returns:
        list[str]: Non-empty text chunks.
    """
    if not isinstance(text, str):
        raise TypeError("text must be a string.")

    if chunk_size <= 0:
        raise ValueError("chunk_size must be greater than zero.")

    if chunk_overlap < 0 or chunk_overlap >= chunk_size:
        raise ValueError(
            "chunk_overlap must be non-negative and smaller than chunk_size."
        )

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        separators=["\n## ", "\n### ", "\n\n", "\n", ". ", " ", ""],
        length_function=len,
    )

    return [
        chunk.strip()
        for chunk in splitter.split_text(text)
        if chunk.strip()
    ]


def chunk_markdown_file(
    file_path,
    chunk_size=1000,
    chunk_overlap=150,
):
    """
    Read a Markdown file and split it into chunks.

    Returns:
        list[dict]: Chunks with source filename and chunk index.
    """
    file_path = Path(file_path)

    if not file_path.exists():
        raise FileNotFoundError(
            f"Markdown file not found: {file_path}"
        )

    text = file_path.read_text(encoding="utf-8")

    chunks = chunk_text(
        text,
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
    )

    return [
        {
            "text": chunk,
            "source": file_path.name,
            "chunk_index": index,
        }
        for index, chunk in enumerate(chunks)
    ]


def chunk_parsed_directory(
    parsed_dir=None,
    chunk_size=1000,
    chunk_overlap=150,
):
    """
    Process all Markdown files in a directory.

    Returns:
        list[dict]: All chunks with source metadata.
    """
    parsed_dir = Path(parsed_dir or PARSED_DIR)

    if not parsed_dir.exists():
        raise FileNotFoundError(
            f"Parsed directory not found: {parsed_dir}"
        )

    all_chunks = []

    for file_path in sorted(parsed_dir.glob("*.md")):
        print(f"Chunking: {file_path.name}")

        chunks = chunk_markdown_file(
            file_path,
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
        )

        all_chunks.extend(chunks)

        print(f"  Chunks created: {len(chunks)}")

    print(f"\nTotal chunks created: {len(all_chunks)}")

    return all_chunks
```

#### embeddings.py file

```python
%%writefile /content/drive/MyDrive/ReCover/src/embeddings.py

"""
Generate semantic text embeddings using Sentence Transformers.
"""

from sentence_transformers import SentenceTransformer


DEFAULT_MODEL = "sentence-transformers/all-MiniLM-L6-v2"


class EmbeddingModel:
    """
    Generate embeddings for document chunks and user queries.
    """

    def __init__(self, model_name=DEFAULT_MODEL):
        self.model_name = model_name
        self.model = SentenceTransformer(model_name)

    def embed_texts(self, texts, batch_size=32):
        """
        Convert multiple texts into embeddings.

        Args:
            texts: List of strings.
            batch_size: Number of texts processed per batch.

        Returns:
            list[list[float]]: Embedding vectors.
        """
        if not texts:
            return []

        if not all(isinstance(text, str) for text in texts):
            raise TypeError("All input texts must be strings.")

        embeddings = self.model.encode(
            texts,
            batch_size=batch_size,
            convert_to_numpy=True,
            normalize_embeddings=True,
            show_progress_bar=len(texts) > 32,
        )

        return embeddings.tolist()

    def embed_query(self, query):
        """
        Generate an embedding for a single user query.
        """
        if not isinstance(query, str) or not query.strip():
            raise ValueError("query must be a non-empty string.")

        return self.model.encode(
            query,
            convert_to_numpy=True,
            normalize_embeddings=True,
        ).tolist()
```
